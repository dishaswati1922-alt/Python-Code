```python
import pandas as pd
import matplotlib.pyplot as plt

```

```python
df=pd.read_excel("/content/Superstore_Sales_Dataset_5000.xlsx")
```

```python
print(df.head())
```

```python
print(df.describe())
```

```python
print(df.isnull().sum())
```

```python
print(df["Sales"].sum())
```

```python
print(df["Profit"].sum())
```

```python
print(df.groupby("Product")["Sales"].sum().sort_values(ascending=False))
```

```python
print(df.groupby("Region")["Sales"].sum())
```

```python
print(df.groupby("Category")["Sales"].sum())
```

```python
df.groupby("Category")["Sales"].sum().plot(kind="bar")
plt.title("Sales by Category")
plt.xlabel("Sales")
plt.ylabel("Category")
plt.show()
```
